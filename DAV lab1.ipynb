```python
import numpy as np
import pandas as pd

marks = np.array([85,72,68,95,70])

print("Original array")
print(marks)

#Access first element
print("First mark: ",marks[0])

#Access last element
print("Last mark: ",marks[-1])

#Slice first 3 elements
print("First three marks: ",marks[0:3])

#Slice elements from index 2 to 4
print("2 to 4 elements:",marks[2:5])

#Access every second element
print("Every second mark: ",marks[::2])

#Reverse the array
print("Reverse:",marks[::-1])

data = np.array([
    [85,78,90],
    [72,80,75],
    [90,88,95],
])
print(data)

#Access first row
print("First student:",data[0])

#Access element at row 0, column 0
print("First student's first subject:",data[0,0])

#Slice first 3 rows
print("First 3 students:",data[0:3])

#Slice rows 1 to 3 from column 1
print(data[1:4,1])

students = {
    "Name" : ["A","B","C","D","E"],
    "Age" : [20,21,19,22,20],
    "Marks" : [85,90,78,88,95]
}

df = pd.DataFrame(students)

print("DataFrame:")
print(df)

print("\nColumning Indexing:")
#Access the Name Column
print(df["Name"])

print("\nRow Indexing:")
#Access the third row
print(df.iloc[2])

print("\nSpecific Element:")
#Access element at row 1, column 2
print(df.iloc[1,2])

print("\nRow Slicing:")
#Slice rows 1 to 3
print(df.iloc[1:4])

print("\nColumn Slicing:")
#Slice first two columns
print(df.iloc[:,0:2])

print("\nRows and Columns Slicing:")
#Slice rows 1 to 3 and columns 0 to 1
print(df.iloc[1:4,0:2])
```
